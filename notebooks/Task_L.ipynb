```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

#L. Interpolation

# Input data
hours = np.array([7, 8, 9, 10, 11, 12, 13])
temperature = np.array([16, 18.5, 21, 23.5, 26, 27.5, 28])

# Linear interpolation
linear = interp1d(hours, temperature, kind='linear')

# Cubic interpolation
cubic = interp1d(hours, temperature, kind='cubic')

# Estimate every 10 minutes = 1/6 hour
ten_minute_times = np.linspace(7, 13, 37)

linear_estimates = linear(ten_minute_times)
cubic_estimates = cubic(ten_minute_times)

target_time = 9.5  # 9:30

print(f"Linear estimate at 9:30: {linear(target_time):.2f} °C")
print(f"Cubic estimate at 9:30: {cubic(target_time):.2f} °C")

# Plot results
plt.figure(figsize=(10, 6))

plt.scatter(hours, temperature, color='red', label='Original data')
plt.plot(ten_minute_times, linear_estimates, label='Linear interpolation')
plt.plot(ten_minute_times, cubic_estimates, label='Cubic interpolation')

plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Interpolation")
plt.legend()
plt.grid(True)
plt.show()
```

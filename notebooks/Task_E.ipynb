```python
import numpy as np

#E. Solution of Systems of Linear Equations

# x = number of Type A servers
# y = number of Type B servers

# Type A uses 5 CPU units and 6 GB memory
# Type B uses 3 CPU units and 10 GB memory
# Total observed usage: 19 CPU units and 42 GB memory

A = np.array([
    [5.0, 3.0],
    [6.0, 10.0]
])

b = np.array([19.0, 42.0])

solution = np.linalg.solve(A, b)
residual = b - A @ solution

print(f"Type A servers: {solution[0]:.0f}")
print(f"Type B servers: {solution[1]:.0f}")
print("Verification A @ solution:", A @ solution)
print("Residual:", residual)
print("Residual norm:", np.linalg.norm(residual))
```
